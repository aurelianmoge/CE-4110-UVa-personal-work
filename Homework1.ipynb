```python
import cvxpy as cvx
```

```python
x1 = cvx.Variable()
x2 = cvx.Variable()
x3 = cvx.Variable()
x4 = cvx.Variable()

obj_fct = -0.01*(x1+x3-300)**2 + 470 - 0.0033*(x2+x4-100)**2 + 33 - 0.064*(x3+x4)
obj = cvx.Maximize(obj_fct)
```

```python
limit = 400

constraint_1 = [x1 + x2 <= limit]

prob_1 = cvx.Problem(obj, constraint_1)
prob_1.solve()

print('Water allocation = %0.2f TAF' % limit)
print('Objective = $%0.2f' % obj.value)

print('x1 = %0.2f TAF of water allocation dedicated to almonds' % x1.value)
print('x2 = %0.2f TAF of water allocation dedicated to alfalfa' % x2.value)
print('x3 = %0.2f TAF of ground water dedicated to almonds' % x3.value)
print('x4 = %0.2f TAF of ground water dedicated to alfalfa' % x4.value)

print(
    'Constraint %d Shadow Price = %.2f'
    % (1, constraint_1[0].dual_value)
)
```
